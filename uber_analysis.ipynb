```python
df2['Location'] = df2['Location'].str.strip().str.title()

df2['City'] = df2['City'].str.strip().str.title()
```

```python
print("Unmatched Pickup IDs:",
      (~df1['PULocationID'].isin(df2['LocationID'])).sum())

print("Unmatched Drop-off IDs:",
      (~df1['DOLocationID'].isin(df2['LocationID'])).sum())
```

```python
pickup_ids = set(df1['PULocationID'])
dropoff_ids = set(df1['DOLocationID'])
location_ids = set(df2['LocationID'])

print("Unmatched Pickup IDs:")
print(sorted(pickup_ids - location_ids))

print("\nUnmatched Drop-off IDs:")
print(sorted(dropoff_ids - location_ids))
```

```python
df1.drop(
    df1[
        (df1['PULocationID'].isin([264, 265])) |
        (df1['DOLocationID'].isin([264, 265]))
    ].index,
    inplace=True
)
```

```python
print("Unmatched Pickup IDs:",
      (~df1['PULocationID'].isin(df2['LocationID'])).sum())

print("Unmatched Drop-off IDs:",
      (~df1['DOLocationID'].isin(df2['LocationID'])).sum())
```

```python
print("Trip Details Shape:", df1.shape)
print("Location Table Shape:", df2.shape)
```

```python
df1 = df1.merge(
    df2[['LocationID', 'Location', 'City']],
    left_on='PULocationID',
    right_on='LocationID',
    how='left'
)

df1.rename(columns={
    'Location': 'Pickup_Location',
    'City': 'Pickup_City'
}, inplace=True)

df1.drop(columns=['LocationID'], inplace=True)
```

```python
df1 = df1.merge(
    df2[['LocationID', 'Location', 'City']],
    left_on='DOLocationID',
    right_on='LocationID',
    how='left'
)

df1.rename(columns={
    'Location': 'Dropoff_Location',
    'City': 'Dropoff_City'
}, inplace=True)

df1.drop(columns=['LocationID'], inplace=True)
```
